```python

import sqlite3
import pandas as pd

conn = sqlite3.connect("Chinook_Sqlite (1).sqlite")

print("Database connected successfully!")
```

```python

query = """
SELECT
    Customer.LastName,
    Customer.FirstName,
    Track.Name AS TrackName,
    Album.Title AS AlbumTitle
FROM Customer
JOIN Invoice
    ON Customer.CustomerId = Invoice.CustomerId
JOIN InvoiceLine
    ON Invoice.InvoiceId = InvoiceLine.InvoiceId
JOIN Track
    ON InvoiceLine.TrackId = Track.TrackId
JOIN Album
    ON Track.AlbumId = Album.AlbumId
ORDER BY
    Customer.LastName,
    Customer.FirstName;
"""

df = pd.read_sql_query(query, conn)

df.head()
```

I connected to the database using the sqlite3 library and wrote a SQL query that joined five tables: Customer, Invoice, InvoiceLine, Track, and Album.

The results were loaded into a pandas DataFrame showing each customer's last name, first name, purchased track, and album title. The information was sorted alphabetically by customer last name and first name.
