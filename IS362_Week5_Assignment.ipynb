# IS 362 – Week 5 Assignment

This assignment uses Python and pandas along with the nycflights13 dataset to answer the following questions:

1. What is the northernmost airport in the United States?
2. What is the easternmost airport in the United States?
3. On February 12th, 2013, which New York area airport had the windiest weather?

I will use pandas to examine the airport and weather data and research my results to check for possible outliers or incorrect values.

```python
import pandas as pd

# Load the airport and weather datasets
airports_url = "https://raw.githubusercontent.com/hadley/nycflights13/master/data-raw/airports.csv"
weather_url = "https://raw.githubusercontent.com/hadley/nycflights13/master/data-raw/weather.csv"

airports = pd.read_csv(airports_url)
weather = pd.read_csv(weather_url)

# Preview the airport data
airports.head()
```

```python
# Find the five airports with the highest latitude
northernmost = airports.sort_values(by="lat", ascending=False).head(5)

northernmost[["faa", "name", "lat", "lon"]]
```

## Question 1: What is the northernmost airport in the United States?

After sorting the airports by latitude, Dillant Hopkins Airport (EEN) appeared first with a latitude of 72.270833. However, this is an error in the dataset because Dillant Hopkins Airport is located in New Hampshire and is not that far north.

The next airport was Wiley Post-Will Rogers Memorial Airport (BRW) in Utqiagvik, Alaska, with a latitude of 71.285446. I verified this result using FAA information, which places the airport at approximately 71°17' N.

Therefore, based on the dataset and my research, the northernmost airport is **Wiley Post-Will Rogers Memorial Airport (BRW) in Alaska**.

```python
# Find the five airports with the highest longitude
easternmost = airports.sort_values(by="lon", ascending=False).head(5)

easternmost[["faa", "name", "lat", "lon"]]
```

## Question 2: What is the easternmost airport in the United States?

I sorted the airport data by longitude and found several values that required additional research. Some airports, including Montgomery Field (MYF) and Deer Valley Municipal Airport (DVT), appeared with positive longitudes even though they are located in the western United States. This indicates errors in the longitude data.

Eareckson Air Station (SYA) in Alaska also appeared because its location is near the 180-degree longitude line, which makes the definition of "easternmost" more complicated.

After reviewing the data and considering the airport locations, I determined that **Eastport Municipal Airport (EPM) in Maine** is the easternmost airport for this analysis. Its longitude in the dataset is approximately -67.012694.

Therefore, my answer is **Eastport Municipal Airport (EPM), Maine**.

```python
# Filter weather data for February 12, 2013
feb12_weather = weather[
    (weather["year"] == 2013) &
    (weather["month"] == 2) &
    (weather["day"] == 12)
]

# Show wind data for the NYC airports
feb12_weather[["origin", "hour", "wind_speed", "wind_gust"]]
```

```python
# Remove unrealistic wind-speed values
valid_weather = feb12_weather[feb12_weather["wind_speed"] < 100]

# Compare the maximum valid wind speed at each airport
valid_weather.groupby("origin")["wind_speed"].max().sort_values(ascending=False)
```

## Question 3: On February 12th, 2013, which New York area airport had the windiest weather?

I filtered the weather data to February 12, 2013 and compared the wind speeds recorded at EWR, JFK, and LGA.

While reviewing the data, I found an unrealistic wind speed of 1048.36058 mph at EWR. I treated this value as an outlier/data error and excluded it from the comparison.

After removing the outlier, the highest wind speeds were:

- LGA: 23.01560 mph
- EWR: 21.86482 mph
- JFK: 20.71404 mph

Therefore, **LaGuardia Airport (LGA)** had the windiest weather on February 12, 2013, with a maximum valid wind speed of approximately **23.02 mph**.
