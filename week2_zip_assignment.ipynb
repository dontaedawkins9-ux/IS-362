# Week 2 Assignment: Using the `zip()` Function in Python

In this notebook, I will explore the Python `zip()` function. I chose `zip()` because combining multiple lists can be confusing at first, especially when the lists have different lengths.

The `zip()` function allows us to combine items from two or more lists based on their position.

## Example 1: Combining Two Lists

In this example, I have one list of student names and another list of scores. I can use `zip()` to match each student with the score in the same position.

```python
students = ["John", "Maria", "David", "Ashley"]
scores = [85, 92, 78, 95]

student_scores = zip(students, scores)

print(list(student_scores))
```

The output is a list of tuples. Each tuple contains one student and the matching score.

## Example 2: Using `zip()` in a Loop

Instead of converting the result to a list, I can also use `zip()` directly inside a `for` loop.

```python
students = ["John", "Maria", "David", "Ashley"]
scores = [85, 92, 78, 95]

for student, score in zip(students, scores):
    print(student, "scored", score)
```

This is useful because Python separates each pair into the variables `student` and `score` automatically.

## Example 3: Creating a Dictionary with `zip()`

Another useful feature of `zip()` is that it can help create a dictionary. The first list can become the keys, and the second list can become the values.

```python
students = ["John", "Maria", "David", "Ashley"]
scores = [85, 92, 78, 95]

score_dictionary = dict(zip(students, scores))

print(score_dictionary)
```

The result is a dictionary where each student's name is connected to that student's score.

## Example 4: What Happens When Lists Have Different Lengths?

One part of `zip()` that can be challenging is understanding what happens when the lists are not the same size. Python stops when the shortest list runs out of items.

```python
names = ["John", "Maria", "David", "Ashley"]
ages = [21, 22, 20]

combined = list(zip(names, ages))

print(combined)
```

Even though the `names` list has four items, only three pairs are created because the `ages` list only contains three items.

## Conclusion

The `zip()` function is useful when working with multiple Python data structures that contain related information. It can combine lists, make loops easier to read, and help create dictionaries.

The part I found most important to remember is that `zip()` stops at the shortest list when the lists have different lengths.
