# IS 362 – Bicycle Trip from NYC to Key West

This notebook uses a pandas Series to store cumulative bicycle odometer readings for a trip from New York City to Key West, Florida. It then calculates the number of miles ridden each day.

```python
import pandas as pd
```

```python
cumulative_miles = pd.Series(
    [55, 120, 185, 250, 315, 380, 445, 510, 575, 640],
    index=['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5',
           'Day 6', 'Day 7', 'Day 8', 'Day 9', 'Day 10']
)

print("Cumulative Miles:")
print(cumulative_miles)
```

## Miles Ridden Each Day

Since the odometer readings are cumulative, subtracting the previous day's reading from the current day's reading gives the number of miles ridden that day. For Day 1, the cumulative reading is also the distance traveled that day.

```python
daily_miles = cumulative_miles.diff()

daily_miles.iloc[0] = cumulative_miles.iloc[0]

print("Miles Ridden Each Day:")
print(daily_miles)
```

## Trip Summary

The table below shows the cumulative odometer reading at the end of each day and the actual number of miles ridden that day.

```python
trip_data = pd.DataFrame({
    'Cumulative Miles': cumulative_miles,
    'Miles Ridden That Day': daily_miles
})

trip_data
```
